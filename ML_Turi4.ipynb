# Baca Data

```python
import pandas as pd

df = pd.read_csv('data_segmentasi_pelanggan.csv')
df
```

# Langkah 1
Menyiapkan data & scaling

```python
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler

# df berisi usia, pendapatan, skor_belanja (tanpa label)
X_scaled = StandardScaler().fit_transform(df)
```

# Langkah 2
Metode Elbow menentukan K

```python
from sklearn.cluster import KMeans
inertias = []
for k in range(1, 11):
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(X_scaled); inertias.append(km.inertia_)
plt.plot(range(1,11), inertias, marker='o'); plt.show()
```

# Langkah 3
Menjalankan K-Means (K=3)

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df['cluster'] = kmeans.fit_predict(X_scaled)
print(df['cluster'].value_counts())
```

# Langkah 4
Evaluasi silhouette score

```python
from sklearn.metrics import silhouette_score
print(silhouette_score(X_scaled, df['cluster']))
```

# Langkah 5
Interpretasi: profil rata-rata tiap cluster

```python
profil = df.groupby('cluster')[['usia','pendapatan',
    'skor_belanja']].mean().round(1)
print(profil)   # beri nama tiap cluster berdasarkan profil
```

# Penamaan Cluster

```python
nama_cluster = {
    0: 'Pelanggan Menengah',
    1: 'Pelanggan Hemat',
    2: 'Pelanggan Aktif Belanja'
}

df['nama_cluster'] = df['cluster'].map(nama_cluster)

print(df)
```

# **LATIHAN MANDIRI**

# Menjalankan K-Means ( K = 5 )

```python
kmeans_5 = KMeans(n_clusters=5, random_state=42, n_init=10)
cluster_5 = kmeans_5.fit_predict(X_scaled)

silhouette_5 = silhouette_score(X_scaled, cluster_5)

print("Silhouette Score K=3:", silhouette_score(X_scaled, df['cluster']))
print("Silhouette Score K=5:", silhouette_5)
```

Hasil pengujian menunjukan, silhouette score pada K=3 sebesar 0,6722, sedangkan pada K=5 sebesar 0,5600. Karena nilai silhouette score K=3 lebih tinggi, maka K=3 memberikan hasil clustering yang lebih baik dibandingkan K=5.

# K-Means Tanpa Scalling

```python
kmeans_tanpa_scaling = KMeans(n_clusters=3, random_state=42, n_init=10)
cluster_tanpa_scaling = kmeans_tanpa_scaling.fit_predict(
    df[['usia', 'pendapatan', 'skor_belanja']]
)

silhouette_tanpa_scaling = silhouette_score(
    df[['usia', 'pendapatan', 'skor_belanja']],
    cluster_tanpa_scaling
)

print("Silhouette Score dengan scaling:", silhouette_score(X_scaled, df['cluster']))
print("Silhouette Score tanpa scaling:", silhouette_tanpa_scaling)
```

Berdasarkan hasil pengujian, silhouette score dengan scaling sebesar 0,6722, sedangkan tanpa scaling sebesar 0,6771. Hasil tersebut menunjukkan bahwa pada data ini, nilai silhouette score tanpa scaling sedikit lebih tinggi. Meskipun begitu, scaling tetap penting karena K-Means mengelompokkan data berdasarkan jarak. Dengan scaling, setiap fitur memiliki skala yang seimbang sehingga tidak ada fitur tertentu yang terlalu memengaruhi proses clustering.

# Scutter Plot Cluster

```python
plt.scatter(
    df['usia'],
    df['skor_belanja'],
    c=df['cluster']
)

plt.xlabel('Usia')
plt.ylabel('Skor Belanja')
plt.title('Scatter Plot Hasil Clustering K-Means')

plt.show()
```

Scatter Plot Menunjukan, tiga kelompok pelanggan yang berbeda. Pelanggan usia muda memiliki skor belanja tinggi, usia sedang memiliki skor belanja sedang, sedangkan usia yang lebih tua memiliki skor belanja rendah.

# REFLEKSI

## Bagaimana K-Means menemukan kelompok tanpa label?

K-Means dapat menemukan kelompok dengan melihat kemiripan antar data berdasarkan jaraknya. Dimana, data yang memiliki karakteristik mirip akan dimasukkan ke dalam cluster yang sama. Proses ini dilakukan berulang sampai terbentuk kelompok yang sesuai.

## Mengapa metode Elbow menunjukkan siku di K=3 pada data ini?

Metode Elbow menunjukkan K=3 karena pada titik tersebut penurunan nilai inertia mulai melambat. Jadi, K=3 dianggap sebagai jumlah cluster yang paling sesuai untuk data ini.

## Mengapa scaling penting, dan apa dampaknya bila dilewatkan?



Scaling penting agar nilai dari setiap fitur memiliki skala yang seimbang. Jika tidak dilakukan scaling, fitur yang nilainya lebih besar bisa lebih dominan dan memengaruhi hasil clustering.
